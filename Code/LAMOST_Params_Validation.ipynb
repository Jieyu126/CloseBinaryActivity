```python
combinedT = pd.read_csv("../Data/CalibrationteffCombined.csv")
# calculate mean and std of logg difference.
cond = (combinedT.teffspec.between(3800, 8000))
meanDiff = np.mean(combinedT[cond].teffsb2 - combinedT[cond].teffspec)
stdDiff = np.std(combinedT[cond].teffsb2 - combinedT[cond].teffspec)
print("Number of SB2 binaries: ", len(combinedT[cond]))
print("meanDiff: ", meanDiff, "stdDiff: ", stdDiff, "median logg error spectral: ", np.median(combinedT[cond].teffspecerr))


fig, ax = plt.subplots(1, 1, figsize=(9, 7))
cond = (combinedT.source == "Xiong+2023")
if len(combinedT[cond]) > 0:
    ax.errorbar(combinedT[cond].teffsb2, combinedT[cond].teffspec, xerr=combinedT[cond].teffsb2err, yerr=combinedT[cond].teffspecerr, fmt='o', color='green', ecolor='k', elinewidth=1, capsize=4, label=r"$\boldsymbol{\rm Xiong\text{+}2023}$")
cond = (combinedT.source == "Southworth+2015")
if len(combinedT[cond]) > 0:
    ax.errorbar(combinedT.teffsb2[cond], combinedT.teffspec[cond], xerr=combinedT.teffsb2err[cond], yerr=combinedT.teffspecerr[cond], fmt='o', color='red', ecolor='k', elinewidth=1, capsize=4, label=r"$\boldsymbol{\rm Southworth\text{+}2015}$")
cond = (combinedT.source == "Torres+2010")
if len(combinedT[cond]) > 0:
    ax.errorbar(combinedT.teffsb2[cond], combinedT.teffspec[cond], xerr=combinedT.teffsb2err[cond], yerr=combinedT.teffspecerr[cond], fmt='o', color='blue', ecolor='k', elinewidth=1, capsize=4, label=r"$\boldsymbol{\rm Torres\text{+}2010}$")
# add mean and std in the figure as text.
ax.annotate(r"$\boldsymbol{\rm a}$", (0.05, 0.93), xycoords='axes fraction')
ax.text(4000, 7300, r"$\boldsymbol{\rm Offset}$" + ": {:.0f}".format(meanDiff) + r"$\boldsymbol{\rm~K}$", fontsize=20)
ax.text(4000, 7000, r"$\boldsymbol{\rm Std}$" + ": {:.0f}".format(stdDiff) + r"$\boldsymbol{\rm~K}$", fontsize=20)
ax.axhline(3.5, color='k', linestyle='--', lw=2)
ax.legend(fontsize=20, loc="lower right")
ax.plot([3000, 9000], [2800, 8800], 'k-.', lw=2)
ax.plot([3000, 9000], [3000, 9000], 'k--', lw=3)
ax.plot([3000, 9000], [3200, 9200], 'k-.', lw=2)

# ax.axhline(7000, color='k', linestyle='--', lw=2)
ax.set_xlim(3800, 8200)
ax.set_ylim(3800, 8200)
ax.set_xlabel(r"$\boldsymbol{T_{\rm eff,~primary}~(\rm K)}$")
ax.set_ylabel(r"$\boldsymbol{T_{\rm eff,~LAMOST}~(\rm K)}$")
plt.savefig("../Figures/teff_comparison.jpg", dpi=300, bbox_inches="tight")
plt.savefig("../Figures/teff_comparison.pdf", bbox_inches="tight")
```

```python
# import matplotlib.pyplot as plt
# import numpy as np
# import pandas as pd

# Read in the data
combinedT = pd.read_csv("../Data/CalibrationteffCombined.csv")
combinedG = pd.read_csv("../Data/CalibrationloggCombined.csv")
combinedM = pd.read_csv("../Data/CalibrationMassCombined.csv")

# Create a new figure
fig = plt.figure(figsize=(17, 12))

# Manually define positions [left, bottom, width, height] in normalized figure coordinates.
# Here, the two top panels each have a width of 0.35:
#   - Left panel: left = 0.1, bottom = 0.55
#   - Right panel: left = 0.55, bottom = 0.55
# The bottom panel (mass) also has a width of 0.35 and is centered (left = 0.325).
ax_teff = fig.add_axes([0.1,   0.55, 0.35, 0.35])  # Top-left (Teff)
ax_logg = fig.add_axes([0.55,  0.55, 0.35, 0.35])  # Top-right (logg)
ax_mass = fig.add_axes([0.325, 0.1,  0.35, 0.35])   # Bottom-center (Mass)

############################################
# Plot 1: Effective Temperature Comparison #
############################################
condT = combinedT.teffspec.between(3800, 8000)
meanDiffT = np.mean(combinedT[condT].teffsb2 - combinedT[condT].teffspec)
stdDiffT = np.std(combinedT[condT].teffsb2 - combinedT[condT].teffspec)
print("Number of SB2 binaries (teff):", len(combinedT[condT]))
print("meanDiff (teff):", meanDiffT, "stdDiff (teff):", stdDiffT, "median teff error spectral:", np.median(combinedT[condT].teffspecerr))

cond = (combinedT.source == "Xiong+2023")
if len(combinedT[cond]) > 0:
    ax_teff.errorbar(combinedT[cond].teffsb2, combinedT[cond].teffspec,
                     xerr=combinedT[cond].teffsb2err, yerr=combinedT[cond].teffspecerr,
                     fmt='o', color='green', ecolor='k', elinewidth=1, capsize=4,
                     label=r"$\boldsymbol{\rm Xiong\text{+}2023}$")
cond = (combinedT.source == "Southworth+2015")
if len(combinedT[cond]) > 0:
    ax_teff.errorbar(combinedT.teffsb2[cond], combinedT.teffspec[cond],
                     xerr=combinedT.teffsb2err[cond], yerr=combinedT.teffspecerr[cond],
                     fmt='o', color='red', ecolor='k', elinewidth=1, capsize=4,
                     label=r"$\boldsymbol{\rm Southworth\text{+}2015}$")
cond = (combinedT.source == "Torres+2010")
if len(combinedT[cond]) > 0:
    ax_teff.errorbar(combinedT.teffsb2[cond], combinedT.teffspec[cond],
                     xerr=combinedT.teffsb2err[cond], yerr=combinedT.teffspecerr[cond],
                     fmt='o', color='blue', ecolor='k', elinewidth=1, capsize=4,
                     label=r"$\boldsymbol{\rm Torres\text{+}2010}$")
ax_teff.annotate(r"$\boldsymbol{\rm a}$", (0.05, 0.93), xycoords='axes fraction')
ax_teff.text(4000, 7300, r"$\boldsymbol{\rm Offset}$" + ": {:.0f}".format(meanDiffT) + r"$\boldsymbol{\rm~K}$", fontsize=20)
ax_teff.text(4000, 7000, r"$\boldsymbol{\rm Std}$" + ": {:.0f}".format(stdDiffT) + r"$\boldsymbol{\rm~K}$", fontsize=20)
ax_teff.axhline(3.5, color='k', linestyle='--', lw=2)
ax_teff.legend(fontsize=20, loc="lower right")
ax_teff.plot([3000, 9000], [2800, 8800], 'k-.', lw=2)
ax_teff.plot([3000, 9000], [3000, 9000], 'k--', lw=3)
ax_teff.plot([3000, 9000], [3200, 9200], 'k-.', lw=2)
ax_teff.set_xlim(3800, 8200)
ax_teff.set_ylim(3800, 8200)
ax_teff.set_xlabel(r"$\boldsymbol{T_{\rm eff,~primary}~(\rm K)}$")
ax_teff.set_ylabel(r"$\boldsymbol{T_{\rm eff,~LAMOST}~(\rm K)}$")

##########################################
# Plot 2: Surface Gravity Comparison (logg)
##########################################
condG = combinedG.loggspec.between(2.5, 4.8)
meanDiffG = np.mean(combinedG[condG].loggsb2 - combinedG[condG].loggspec)
stdDiffG = np.std(combinedG[condG].loggsb2 - combinedG[condG].loggspec)
print("Number of SB2 binaries (logg):", len(combinedG[condG]))
print("meanDiff (logg):", meanDiffG, "stdDiff (logg):", stdDiffG, "median logg error spectral:", np.median(combinedG[condG].loggspecerr))

cond = (combinedG.source == "Xiong+2023")
if len(combinedG[cond]) > 0:
    ax_logg.errorbar(combinedG[cond].loggsb2, combinedG[cond].loggspec,
                     xerr=combinedG[cond].loggsb2err, yerr=combinedG[cond].loggspecerr,
                     fmt='o', color='green', ecolor='k', elinewidth=1, capsize=4,
                     label=r"$\boldsymbol{\rm Xiong\text{+}2023}$")
cond = (combinedG.source == "Southworth+2015")
if len(combinedG[cond]) > 0:
    ax_logg.errorbar(combinedG.loggsb2[cond], combinedG.loggspec[cond],
                     xerr=combinedG.loggsb2err[cond], yerr=combinedG.loggspecerr[cond],
                     fmt='o', color='red', ecolor='k', elinewidth=1, capsize=4,
                     label=r"$\boldsymbol{\rm Southworth\text{+}2015}$")
cond = (combinedG.source == "Torres+2010")
if len(combinedG[cond]) > 0:
    ax_logg.errorbar(combinedG.loggsb2[cond], combinedG.loggspec[cond],
                     xerr=combinedG.loggsb2err[cond], yerr=combinedG.loggspecerr[cond],
                     fmt='o', color='blue', ecolor='k', elinewidth=1, capsize=4,
                     label=r"$\boldsymbol{\rm Torres\text{+}2010}$")
ax_logg.text(2.6, 4.35, r"$\boldsymbol{\rm Offset}$" + ": {:.2f}".format(meanDiffG) + r"$\boldsymbol{\rm~dex}$", fontsize=20)
ax_logg.text(2.6, 4.21, r"$\boldsymbol{\rm Std}$" + ": {:.2f}".format(stdDiffG) + r"$\boldsymbol{\rm~dex}$", fontsize=20)
ax_logg.legend(fontsize=20, loc="lower right")
ax_logg.plot([2.0, 4.8], [2.0, 4.8], 'k--', lw=3)
ax_logg.plot([2.0, 4.8], [1.9, 4.7], 'k-.', lw=2)
ax_logg.plot([2.0, 4.8], [2.1, 4.9], 'k-.', lw=2)
ax_logg.set_xlim(2.5, 4.8)
ax_logg.set_ylim(2.5, 4.8)
ax_logg.set_xlabel(r"$\boldsymbol{\log{g}_{\rm~primary}}$")
ax_logg.set_ylabel(r"$\boldsymbol{\log{g}_{\rm~LAMOST}}$")
ax_logg.annotate(r"$\boldsymbol{\rm b}$", (0.05, 0.93), xycoords='axes fraction')

######################################
# Plot 3: Mass Comparison (bottom row)
######################################
condM = (combinedM.masssb2.between(0., 5.5) & combinedM.iso_mass.between(0.5, 5.5))
meanDiffM = np.mean(combinedM[condM].masssb2 - combinedM[condM].iso_mass)
stdDiffM = np.std(combinedM[condM].masssb2 - combinedM[condM].iso_mass)
print("Number of SB2 binaries (mass):", len(combinedM[condM]))
print("meanDiff (mass):", meanDiffM, "stdDiff (mass):", stdDiffM, "median mass:", np.median(combinedM[condM].masssb2))
print("median mass error:", np.median(combinedM[condM].iso_mass_err))
print("median relative error in mass:", np.median(np.std(np.abs(combinedM[condM].masssb2 - combinedM[condM].iso_mass) / combinedM[condM].masssb2)))

cond = (combinedM.source == "Xiong+2023")
if len(combinedM[cond]) > 0:
    ax_mass.errorbar(combinedM[cond].masssb2, combinedM[cond].iso_mass,
                     xerr=combinedM[cond].masssb2err, yerr=combinedM[cond].iso_mass_err,
                     fmt='o', color='green', ecolor='k', elinewidth=1, capsize=4,
                     label=r"$\boldsymbol{\rm Xiong\text{+}2023}$")
cond = (combinedM.source == "Southworth+2015")
if len(combinedM[cond]) > 0:
    ax_mass.errorbar(combinedM.masssb2[cond], combinedM.iso_mass[cond],
                     xerr=combinedM.masssb2err[cond], yerr=combinedM.iso_mass_err[cond],
                     fmt='o', color='red', ecolor='k', elinewidth=1, capsize=4,
                     label=r"$\boldsymbol{\rm Southworth\text{+}2015}$")
cond = (combinedM.source == "Torres+2010")
if len(combinedM[cond]) > 0:
    ax_mass.errorbar(combinedM.masssb2[cond], combinedM.iso_mass[cond],
                     xerr=combinedM.masssb2err[cond], yerr=combinedM.iso_mass_err[cond],
                     fmt='o', color='blue', ecolor='k', elinewidth=1, capsize=4,
                     label=r"$\boldsymbol{\rm Torres\text{+}2010}$")
ax_mass.annotate(r"$\boldsymbol{\rm c}$", (0.05, 0.93), xycoords='axes fraction')
ax_mass.text(0.4, 2.5, r"$\boldsymbol{\rm Offset}$" + ": {:.2f}".format(meanDiffM) + r"$\boldsymbol{\rm~M_{\odot}}$", fontsize=20)
ax_mass.text(0.4, 2.32, r"$\boldsymbol{\rm Std}$" + ": {:.2f}".format(stdDiffM) + r"$\boldsymbol{\rm~M_{\odot}}$", fontsize=20)
ax_mass.legend(fontsize=20, loc="lower right")
ax_mass.plot([0, 3.2], [0, 3.2], 'k--', lw=3)
ax_mass.plot([0, 3.2], [0.2, 3.4], 'k-.', lw=2)
ax_mass.plot([0, 3.2], [-0.2, 3.0], 'k-.', lw=2)
ax_mass.set_xlim(0.3, 3.)
ax_mass.set_ylim(0.3, 3.)
ax_mass.set_xlabel(r"$\boldsymbol{\rm~Mass_{~primary}~(\rm M_{\odot})}$")
ax_mass.set_ylabel(r"$\boldsymbol{\rm~Mass_{~LAMOST}~(\rm M_{\odot})}$")

# Save and show the combined figure
plt.savefig("../Figures/paramsValidation.jpg", dpi=300, bbox_inches="tight")
plt.savefig("../Figures/paramsValidation.pdf", bbox_inches="tight")
```
